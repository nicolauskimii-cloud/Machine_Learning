```python
import pandas as pd
import numpy as np

from sklearn.decomposition import PCA

x = pd.read_excel('nba_baru.xlsx')  # Load your data from an Excel file
x
```

```python
features1 = ["Height", "Weight"]
features2 = ["Age", "Salary"]

data1 = x[features1].copy()  # Separating out the features
data2 = x[features2].copy()  # Separating out the target variable
data1
data2
```

```python
import matplotlib.pyplot as plt
plt.scatter(data1.iloc[:,0],data1.iloc[:,1])
plt.show()
```

```python
plt.scatter(data2.iloc[:,0],data2.iloc[:,1])
plt.show()
```

```python
pca1 = PCA(n_components=2)
pca_result1 = pca1.fit_transform(data1)
```

```python
print('Matrix covarian')
print(pca1.get_covariance())
```

```python
pca2 = PCA(n_components=2)
pca_result2 = pca2.fit_transform(data2)
```

```python
print('Matrix covarian')
print(pca2.get_covariance())
```

```python
print("Eigenvalue : ")
print(pca1.explained_variance_)
```

```python
print("Eigenvalue : ")
print(pca2.explained_variance_)
```

```python
print("Eigenvector : ")
print(pca1.components_)
```

```python
print("Eigenvector : ")
print(pca2.components_)
```

```python
print("Hasil Transformasi PCA 1")
print(pca_result1)
```

```python
print("Hasil Transformasi PCA 2")
print(pca_result2)
```

```python
plt.scatter(pca_result1[:,0],pca_result1[:,1])
plt.show()
```

```python
plt.scatter(pca_result2[:,0],pca_result2[:,1])
plt.show()
```

```python
#Tahap 1, Variance Thrfeshold
variance1 = data1.var()
print("Variance : ",variance1.sort_values(ascending=False))
```

```python
#means
means1 = data1.mean()
print("Mean : ",means1.sort_values(ascending=False))
```

```python
#Tahap 2, Correlation Analysis
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
plt.figure(figsize=(12,8))

corr = data1.corr()
corr
```

```python
from sklearn.feature_selection import VarianceThreshold
selector = VarianceThreshold(15)
selector.fit_transform(data1)
```

```python
df_pca = pd.DataFrame(data = pca_result1, columns = ['PC1', 'PC2'])
print("Korelasi antar komponen")
print(df_pca.corr())
```

```python
sns.heatmap(df_pca.corr(), annot=True, cmap=plt.cm.Reds)
plt.show()
```
