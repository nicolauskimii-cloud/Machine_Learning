```python
import pandas as pd
import numpy as np

#importing the dataset 
df = pd.read_excel('datagaji.xlsx')
df.info()
```

```python
dfupdatena = df.copy()
dfupdatena.head
```

```python
dfupdatena['USIA']= dfupdatena['USIA'].fillna(dfupdatena['USIA'].mean())
dfupdatena.info()
```

```python
dfupdatena.head
```

```python
dfupdatena['GAJI'] = dfupdatena['GAJI'].fillna(dfupdatena['GAJI'].mean())
dfupdatena.info()
```

```python
dfupdatena
```

```python
dfupdatena.describe()
```

```python
dfupdatezscore = dfupdatena.copy()
dfupdatezscore
```

```python
dfupdatezscore.drop("NEGARA",axis=1,inplace=True)
dfupdatezscore.drop("BELI",axis=1,inplace=True)
```

```python
dfupdatezscore
```

```python
x = dfupdatena[['USIA', 'GAJI']]
x
```

```python
from sklearn.preprocessing import StandardScaler

ss =StandardScaler()
zscore_df = ss.fit_transform(dfupdatezscore)
```

```python
zscore_df
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
x_minmax = scaler.fit_transform(dfupdatezscore)
x_minmax
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler2 = MinMaxScaler(feature_range=(1,10))
x_minmax2 = scaler2.fit_transform(dfupdatezscore)
x_minmax2
```

```python
df_minmax2 = pd.DataFrame(x_minmax2, columns=['MinMax10_USIA','MinMax10_GAJI' ])
df_minmax2
```

```python
concatenated3 = pd.concat([dfupdatena, df_minmax2], axis="columns")
concatenated3
```

```python
df_zscore=pd.DataFrame(zscore_df, columns=['ZSCORE_USIA', 'ZSCORE_GAJI'])
df_minmax1 = pd.DataFrame(x_minmax, columns=['MinMax01_USIA', 'MinMax01_GAJI'])
df_minmax2 = pd.DataFrame(x_minmax2, columns=['MinMax10_USIA', 'MinMax10_GAJI']) 
```

```python
gabung_all=pd.concat([dfupdatena,df_zscore,df_minmax2,df_minmax1], axis="columns")
gabung_all
```
