```python
import pandas as pd
import numpy as np

df = pd.read_csv("all_seasons.csv")
df

```

```python
df = df.drop(columns="Unnamed: 0")
```

```python
df.info()
```

```python
df
```

```python
target = "pts"

features = df.select_dtypes(include="number").columns
features = features[features != target]
    
```

```python
x = df[features]
y = df[target]

variance = x.var()

print("\n=== TAHAP 1: VARIANCE ===")
print(variance.sort_values(ascending=False))
```

```python
print("\n=== TAHAP 1: MEAN ===")
mean = x.mean()
print(mean.sort_values(ascending=False))
```

```python
from sklearn.feature_selection import VarianceThreshold
selector = VarianceThreshold(15)
selector.fit_transform(x)
```

```python
# ============================================
# TAHAP 2
# CORRELATION ANALYSIS
# ============================================
## Pearson correlation

import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
plt.figure(figsize=(12, 10))
corr = x.corr()
print("\n=== TAHAP 2: CORRELATION ===")
sns.heatmap(corr, annot=True, cmap=plt.cm.Reds)
plt.show()
```

```python
# ============================================
# TAHAP 2B
# CORRELATION ANALYSIS
# ============================================

correlation=x.corrwith(y)
correlation_result = pd.DataFrame({
    "Feature": features,
    "Pearson_r": correlation.values,
    "Abs_r": correlation.abs().values
})

correlation_result = correlation_result.sort_values(
    "Abs_r",
    ascending=False
)

print("\n=== TAHAP 2: CORRELATION ===")
print(correlation_result)
```

```python
# ============================================
# TAHAP 3
# UNIVARIATE F-TEST
# ============================================
from sklearn.feature_selection import f_regression
F, p = f_regression(x, y)

univariate_result = pd.DataFrame({
    "Feature": features,
    "F_statistic": F,
    "p_value": p
})

univariate_result = univariate_result.sort_values(
    "F_statistic",
    ascending=False
)

univariate_result["Significant"] = (
    univariate_result["p_value"] < 0.05
)

print("\n=== TAHAP 3: UNIVARIATE ===")
print(univariate_result)
```

```python
# ============================================
# TAHAP 4
# FINAL RANKING
# ============================================

final_result = correlation_result.merge(
    univariate_result,
    on="Feature"
)

final_result = final_result.sort_values(
    "Abs_r",
    ascending=False
)

final_result["Final_Rank"] = range(
    1,
    len(final_result) + 1
)

print("\n=== TAHAP 4: FINAL RANKING ===")
print(final_result)
```

```python
# ============================================
# FINAL FEATURE SELECTION
# ============================================

selected_features = final_result.loc[
    final_result["Significant"],
    "Feature"
].tolist()

print("\n=== FITUR TERPILIH ===")
print(selected_features)
```
